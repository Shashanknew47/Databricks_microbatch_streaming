### Back pressure: 
- Lets says we have to refresh a dashboard in every 15 mins. There are 10 million records to update in every 15 mins. 
- Here you will have to do Cluser 
  1. Startup time, 
  2. Processing time and Volume
  3. Clean up / Housekeeping time +
- As your Data will be bigger processing this 15 mins will be difficult to process. Lets say this time you got 12 million records and they are getting processed in 17 mins. In every cycle every iteration is getting delayed by 2 mins and few hours this delay will be few hours. This delay is called `Back Pressure`. 



### Incremental Data processing (CheckPoint):
- As our Data size grows we need to pull only those records which are new in the source DB. 
- When we records get processed first time it records which records were processed. So, next time it will process only new records. This process is called check pointing.
